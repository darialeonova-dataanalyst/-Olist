```python
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.stats.proportion import confint_proportions_2indep

df = pd.read_csv("final_result.csv")

# Создаем разметку для теста
df['is_late'] = df['delay_days'] > 0
df['low_rating'] = df['review_score'].isin([1, 2])

# Строим таблицу сопряженности
table = pd.crosstab(df["is_late"], df["low_rating"])
print("--- Таблица сопряжённости ---")
print(table)
print("-" * 30)

# Проводим тест Хи-квадрат
chi2, p_value, dof, expected = chi2_contingency(table)
print(f"Chi-square: {chi2:.3f}")
print(f"p-value: {p_value:.6g}")

# Считаем доверительный интервал для разницы долей
neg_late = table.loc[True, True]
total_late = table.loc[True].sum()

neg_ontime = table.loc[False, True]
total_ontime = table.loc[False].sum()

p1 = neg_late / total_late
p2 = neg_ontime / total_ontime
diff = p1 - p2

# Метод 'wald' — стандартный классический метод расчета
ci_low, ci_upp = confint_proportions_2indep(neg_late, total_late, neg_ontime, total_ontime, method='wald', alpha=0.05)

print("-" * 30)
print(f"Доля негатива при опоздании: {p1*100:.2f}%")
print(f"Доля негатива вовремя: {p2*100:.2f}%")
print(f"Разница долей: {diff*100:.2f} п.п.")
print(f"95%-й доверительный интервал для разницы долей: [{ci_low*100:.2f} п.п., {ci_upp*100:.2f} п.п.]")
```
